```python
import heapq
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
```

```python
TAGS = [10, 50, 100, 200, 400]
FILE_PATHS = [f"../ablation/t_min_{m}_unit_400/110/r_term.txt" for m in TAGS]
```

```python
def parse_data():
    combined = []
    for filepath in FILE_PATHS:
        result = []
        with open(filepath, 'r') as f:
            for line in f:
                result.append(float(line.strip()))
        combined.append(result)
    return combined
```

```python
def draw_boxplot(data: list[list[float]]):
    fig, ax = plt.subplots()

    ax.boxplot(data, tick_labels=[str(v) for v in TAGS], patch_artist=True)

    ax.set_xlabel("I_min (mS)")
    ax.set_ylabel("terminal reachability (%)")

    plt.tight_layout()
    plt.show()
```

```python
draw_boxplot(parse_data())
```
